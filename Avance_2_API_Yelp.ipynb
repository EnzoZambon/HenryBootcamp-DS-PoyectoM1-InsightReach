Relleno de nulos en la columna precio

```python
#El proceso de asignación de valor consiste en crear una columna nueva asignando a cada $ un valor numerico

df_rest['price_num'] = df_rest['price'].map({
    '$': 1,
    '$$': 2,
    '$$$': 3,
    '$$$$': 4
}).astype('Int64')

df_rest['price_num']
```

```python
#Para aquellos precios que quedaron sin asignarles un valor, buscaremos asignarle uno a partir de información de otras columnas. En este caso, 
# tomamos como variable de referencia "categories" para internar estimar el valor del precio de los restaurantes
#EL problema es que las categorias se encuentran en un formato de lista de diccionarios, con lo cual debemos hacer un proceso de extraccion de las mismas, 
# Creamos una nueva columna, llamada 'category_title' donde le asginamos a cada una el valor de acuerdo a su restaurante
df_exploded = df_rest.explode('categories').reset_index(drop=True)
df_exploded['category_title'] = df_exploded['categories'].apply(lambda x: x.get('title') if isinstance(x, dict) else None)
df_exploded.head()
```

```python
#Posteriormente generamos una lista de precios para cada categoria en función del promedio de las que aparecen del total de restaurantes
df_category_prices = df_exploded.groupby('category_title', as_index=False)['price_num'].mean()
df_category_prices['price_num'] = df_category_prices['price_num'].astype('Int64')
df_category_prices
```

```python
#Finalmente, asginamos estos resultados en función de las categorias que tenga cada restaurante cuyo precio sea NAN

df_exp = df_rest.explode('categories').reset_index(drop=True)

# Extraer el título de cada categoría
df_exp['category_title'] = df_exp['categories'].apply(lambda x: x['title'] if isinstance(x, dict) else None)

# Merge con df_category_prices para traer el promedio de precio por categoría
df_exp = df_exp.merge(
    df_category_prices.rename(columns={'price_num': 'cat_avg_price'}),
    on='category_title',
    how='left'
)

# Imputar NaN de price_num con cat_avg_price
df_exp['price_num'] = df_exp['price_num'].fillna(df_exp['cat_avg_price'])

# Volver a reagrupar por restaurante
df_rest['price_num'] = df_exp.groupby(df_exp.index)['price_num'].mean()

#Convertir a Int64 para manejar nulos
df_rest['price_num'] = df_rest['price_num'].astype('Int64')

df_rest['price_num']
```

CREACIÓN DE TABLAS Y ANALISIS DE LA INFORMACIÓN

1. Comenzamos por armar un Promedio Ponderado de los restaurantes para tener una medida mas precisa acorde a las reviews y el rating

```python
# 1. Calcular promedio global de rating
C = df_rest['rating'].mean()

# 2. Definir mínimo de reviews para considerar (puede ser percentil 70)
m = df_rest['review_count'].quantile(0.80)

# 3. Función para calcular Weighted Rating
def weighted_rating(x, m=m, C=C):
    v = x['review_count']
    R = x['rating']
    return (v / (v + m)) * R + (m / (v + m)) * C

# 4. Crear nueva columna con Weighted Rating
df_rest['weighted_rating'] = df_rest.apply(weighted_rating, axis=1).round(2)

# 5. Ordenar de mayor a menor
df_rest_sorted = df_rest.sort_values('weighted_rating', ascending=False)

# Ver top 10 restaurantes
df_rest_sorted[['name', 'rating', 'review_count', 'weighted_rating']].head(10)
```

Vinculamos al promedio ponderado con la categoria principal del restaurante

```python
import ast

# --- 1. Procesar la columna categories ---
# Si la columna 'categories' está como string, convertirla a lista de diccionarios
df_rest['categories'] = df_rest['categories'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

# Crear nueva columna con títulos de categorías separados por coma
df_rest['categories_clean'] = df_rest['categories'].apply(lambda x: ', '.join([d['title'] for d in x]) if isinstance(x, list) else '')

# --- 2. Calcular promedio global de rating ---
C = df_rest['rating'].mean()

# --- 3. Definir mínimo de reviews (percentil 80 por ejemplo) ---
m = df_rest['review_count'].quantile(0.80)

# --- 4. Función Weighted Rating ---
def weighted_rating(x, m=m, C=C):
    v = x['review_count']
    R = x['rating']
    return (v / (v + m)) * R + (m / (v + m)) * C

# --- 5. Crear columna con Weighted Rating ---
df_rest['weighted_rating'] = df_rest.apply(weighted_rating, axis=1).round(2)

# --- 6. Ordenar ---
df_rest_sorted = df_rest.sort_values('weighted_rating', ascending=False)

# --- 7. Ver top 10 incluyendo categorías ---
df_rest_sorted[['name', 'categories_clean', 'rating', 'review_count', 'weighted_rating']].head(10)
```
